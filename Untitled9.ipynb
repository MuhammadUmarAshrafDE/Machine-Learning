```python
import numpy as pd
import matplotlib.pyplot as plt
import pandas as pd

data = pd.read_csv('/content/Housing.csv')
```

```python
data.head()
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.isnull().sum()
```

```python
data.fillna(0)
```

```python
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1
print(IQR)
```

```python
variance = data.var()
print(variance)
```

```python
SD = data.std()
print(SD)
```

```python
data.duplicated().sum()
```

```python
data.drop.duplicates(inplace=True)
```

```python
col = data ['price']
```

```python
#calculate measures like mean, median, sd, iqr variance
mean = col.mean()
median = col.median()
sd = col.std()
iqr = col.quantile(0.75) - col.quantile(0.25)
variance = col.var()
```

```python
#create plots
fig, axs = plt.subplots(1, 5, figsize=(15, 5))
axs[0].bar(['Mean'],[mean])
axs[0].set_title(f'Mean\n {mean:.2f}')
axs[1].bar(['Median'],[median])
axs[1].set_title(f'Median\n{median:.2f}')
axs[4].bar(['Variance'],[variance])
axs[4].set_title(f'Variance\n{variance:.2f}')
axs[2].bar(['Std Dev'],[sd])
axs[2].set_title(f'Std Dev\n{sd:.2f}')
axs[3].bar(['IQR'],[iqr])
axs[3].set_title(f'IQR\n{iqr:.2f}')

plt.suptitle('Descriptive Statistics of Price')
plt.tight_layout()
plt.show()
```
